1. What is the difference between a Basic Filter, Advanced Filter, and a Slicer in Power BI?

Basic Filter: Checkbox-style filter in the Filter Pane to select/deselect specific values.   Advanced Filter: Rule-based filter in the Filter Pane using logical conditions (e.g., Contains, Starts with, Greater than).   Slicer: On-canvas interactive visual that allows report users to filter data directly on the page.   

2. Using the Sales dataset, apply a Basic Filter to display only the sales data for the year 2023.

Select the visual/page.   In the Filters Pane, drag the Date field into Filters on this page (or visual).   Change filter type to Basic filtering.   Expand the years list and check 2023.

4. What is a Visual-Level Filter, and how does it differ from Page-Level and Report-Level Filters?

Visual-Level Filter: Applies only to a single specific visual on the canvas.   Page-Level Filter: Applies to all visuals on the currently selected report page.   Report-Level Filter: Applies globally across every single page in the entire report.

6. What is the role of Drill-Down in Power BI? How does it improve data exploration?

Drill-down allows users to move from a high-level summary view (e.g., Year) down to a granular detail view (e.g., Month or Day) within the same visual, making it easy to spot root causes and deeper trends.

10. Explain the difference between Drill-Down and Drill-Through in Power BI using a real-world example.

Drill-Down: Unpacks levels within the same visual on the same page.

Example: Clicking on "Electronics" expands to show "Laptop, Phone, Tablet" inside the same chart.   

Drill-Through: Navigates from a summary visual to a completely separate detail page carrying context along.

Example: Right-clicking "Laptop" navigates to a dedicated detail page displaying store-wise inventory and customer details for laptops.

13. What is the “Keep All Filters” option in Drill-Through? How does it affect user interaction

When toggled ON, all active slicers, page-level, and report-level filters applied on the source page pass through to the destination page. When OFF, only the specific field selected for drill-through is carried over.
